```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_excel("Telco_Customer_churn.xlsx")
df.head(10)
```

```python
df.info()
```

```python
df[df['TotalCharges'].str.strip() == '']
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(0)
```

```python
df.groupby("Churn")[["tenure", "MonthlyCharges"]].agg(["mean", "median", "std"])
```

```python
from scipy import stats

churned = df[df["Churn"] == "Yes"]["MonthlyCharges"]
stayed = df[df["Churn"] == "No"]["MonthlyCharges"]

stats.ttest_ind(churned, stayed)
```

```python
table = pd.crosstab(df["Contract"], df["Churn"])
print(table)

stats.chi2_contingency(table)
```

```python
df["churn_num"] = (df["Churn"] == "Yes").astype(int)

df[["tenure", "MonthlyCharges", "TotalCharges", "churn_num"]].corr()
```
